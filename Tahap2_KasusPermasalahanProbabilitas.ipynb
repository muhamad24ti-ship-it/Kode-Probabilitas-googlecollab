```python
import math

n = 10
p_gagal = 0.1

p_semua_sukses = (1 - p_gagal) ** n
p_min1_gagal = 1 - p_semua_sukses

# Versi eksak dengan bilangan bulat
pembilang = 10**10 - 9**10
penyebut = 10**10

print(f"P(semua sukses)    : {p_semua_sukses:.10f}")
print(f"P(minimal 1 gagal) : {p_min1_gagal:.10f}")
print(f"Eksak              : {pembilang}/{penyebut}")
print("Cocok dengan hitungan eksak?", math.isclose(p_min1_gagal, pembilang / penyebut))
```

```python
from fractions import Fraction

total = 200
spam, malware, keduanya = 30, 20, 8

p_s = Fraction(spam, total)
p_m = Fraction(malware, total)
p_sm = Fraction(keduanya, total)

p_union = p_s + p_m - p_sm
p_aman = 1 - p_union

print("P(S), P(M), P(S∩M):", p_s, p_m, p_sm)
print("P(spam atau malware):", p_union, "=", float(p_union))
print("P(aman):", p_aman, "=", float(p_aman))
```

```python
import math
from fractions import Fraction

backend, frontend, pilih = 7, 5, 5

total = math.comb(backend + frontend, pilih)
kasus = math.comb(backend, 3) * math.comb(frontend, 2)
p = Fraction(kasus, total)

print("Total kombinasi   :", total)
print("Kasus diinginkan  :", kasus)
print("Peluang (pecahan) :", p)
print("Peluang (desimal) :", round(float(p), 4))
```

```python
import itertools
import math
from fractions import Fraction

ruang_sampel = list(itertools.product(range(1, 6), repeat=3))
digit_beda = [pin for pin in ruang_sampel if len(set(pin)) == 3]

p = Fraction(len(digit_beda), len(ruang_sampel))

print("Jumlah ruang sampel :", len(ruang_sampel))
print("Jumlah digit beda   :", len(digit_beda))
print("Peluang             :", p, "=", float(p))
print("Cek math.perm(5,3)  :", math.perm(5, 3))
print("3 contoh pertama    :", digit_beda[:3])
```

```python
from fractions import Fraction

p_s = Fraction(2, 5)
p_bukan = 1 - p_s
p_g_s = Fraction(1, 2)
p_g_bukan = Fraction(1, 20)

p_g = p_g_s * p_s + p_g_bukan * p_bukan
p_s_g = (p_g_s * p_s) / p_g

print("P(G)   :", p_g)
print("P(S|G) :", p_s_g, "=", round(float(p_s_g), 4))
```

```python
from sympy import symbols, Rational

p, s, f = symbols('p s f')   # p = prevalensi, s = sensitivitas, f = false positive

p_alarm = p*s + (1 - p)*f
posterior = p*s / p_alarm

nilai = {p: Rational(1, 100), s: Rational(19, 20), f: Rational(1, 50)}

print("Rumus simbolik P(M|A):", posterior)
print("P(A)   :", p_alarm.subs(nilai))
print("P(M|A) :", posterior.subs(nilai), "=", round(float(posterior.subs(nilai)), 4))
```

```python
import random
import math

random.seed(42)

N = 100_000   # jumlah percobaan
K = 10        # jumlah request
S = 50        # jumlah server

tabrakan = 0
for _ in range(N):
    pilihan = [random.randrange(S) for _ in range(K)]
    if len(set(pilihan)) < K:     # ada server yang dipilih lebih dari sekali
        tabrakan += 1

p_simulasi = tabrakan / N
p_eksak = 1 - math.perm(S, K) / S**K

print(f"Eksak     : {p_eksak:.4f}")
print(f"Simulasi  : {p_simulasi:.4f}")
print(f"Selisih   : {abs(p_simulasi - p_eksak):.4f}")
```

```python
import numpy as np

rng = np.random.default_rng(42)
N = 1_000_000

lb = rng.random(N) < 0.99        # load balancer hidup
s1 = rng.random(N) < 0.90        # server 1 hidup
s2 = rng.random(N) < 0.90        # server 2 hidup

aktif = lb & (s1 | s2)           # seri dengan paralel

p_simulasi = aktif.mean()
p_eksak = 0.99 * (1 - 0.1**2)

print(f"Eksak    : {p_eksak:.4f}")
print(f"Simulasi : {p_simulasi:.4f}")
print(f"Selisih  : {abs(p_simulasi - p_eksak):.5f}")
```
