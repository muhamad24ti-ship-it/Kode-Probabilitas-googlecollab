```python
import math, random, itertools
from fractions import Fraction
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sympy import symbols, integrate

# 1. math
print("math.comb(5,2)       :", math.comb(5, 2))
print("math.factorial(5)    :", math.factorial(5))

# 2. random
random.seed(42)
print("random.sample (3)    :", random.sample(range(1, 50), 3))

# 3. itertools
print("itertools.combinations:", len(list(itertools.combinations(range(5), 2))))

# 4. fractions
print("fractions            :", Fraction(1, 3) + Fraction(1, 6))

# 5. numpy
print("numpy.mean           :", np.mean([1, 2, 3, 4, 5, 6]))

# 6. scipy.stats
print("scipy binom.pmf      :", stats.binom.pmf(2, 5, 0.5))

# 9. pandas
print("pandas value_counts  :")
print(pd.Series(['A', 'B', 'A', 'A']).value_counts(normalize=True))

# 10. sympy
x = symbols('x')
print("sympy integrate      :", integrate(2*x, (x, 0, 1)))

# 7 & 8. matplotlib + seaborn
fig, ax = plt.subplots(1, 2, figsize=(9, 3))
ax[0].bar([0, 1], [0.5, 0.5])
ax[0].set_title("matplotlib: bar")
sns.histplot([1, 2, 2, 3, 3, 3], kde=True, ax=ax[1])
ax[1].set_title("seaborn: histplot + KDE")
plt.tight_layout()
plt.savefig("tahap1_library.png", dpi=150)
plt.show()
```
