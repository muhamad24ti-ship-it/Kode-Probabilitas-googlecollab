```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

n, p = 10, 0.3
X = stats.binom(n, p)

print("P(X=3)  :", round(X.pmf(3), 4))
print("P(X<=3) :", round(X.cdf(3), 4))
print("P(X>=5) :", round(X.sf(4), 4))      # sf(4) = P(X>4) = P(X>=5)
print("Mean    :", round(X.mean(), 4))
print("Varians :", round(X.var(), 4))

# Simulasi
sim = X.rvs(size=10000, random_state=42)
print("Mean simulasi    :", round(sim.mean(), 3))
print("Varians simulasi :", round(sim.var(), 3))

# Visualisasi
k = np.arange(0, n + 1)
frek_sim = np.array([(sim == i).mean() for i in k])

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
lebar = 0.4
ax[0].bar(k - lebar/2, X.pmf(k), width=lebar, label="Teori (pmf)", color="steelblue")
ax[0].bar(k + lebar/2, frek_sim, width=lebar, label="Simulasi", color="orange")
ax[0].set_title("Binomial(n=10, p=0.3): pmf vs simulasi")
ax[0].set_xlabel("k (jumlah modul berbug)")
ax[0].set_ylabel("Peluang")
ax[0].legend()

ax[1].step(k, X.cdf(k), where="post", color="green")
ax[1].set_title("CDF Binomial(10, 0.3)")
ax[1].set_xlabel("k")
ax[1].set_ylabel("P(X <= k)")

plt.tight_layout()
plt.savefig("tahap3_binomial.png", dpi=150)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

lam = 4
X = stats.poisson(mu=lam)

print("P(X=6)         :", round(X.pmf(6), 4))
print("P(X<=6)        :", round(X.cdf(6), 4))
print("P(overload X>6):", round(X.sf(6), 4))
print("Mean           :", X.mean())
print("Varians        :", X.var())

sim = X.rvs(size=10000, random_state=42)
print("Mean simulasi    :", round(sim.mean(), 3))
print("Varians simulasi :", round(sim.var(), 3))
print("Proporsi overload simulasi:", round((sim > 6).mean(), 4))

k = np.arange(0, 15)
frek_sim = np.array([(sim == i).mean() for i in k])

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
lebar = 0.4
ax[0].bar(k - lebar/2, X.pmf(k), width=lebar, label="Teori (pmf)", color="steelblue")
ax[0].bar(k + lebar/2, frek_sim, width=lebar, label="Simulasi", color="orange")
ax[0].set_title("Poisson(λ=4): pmf vs simulasi")
ax[0].set_xlabel("k (request per detik)")
ax[0].set_ylabel("Peluang")
ax[0].legend()

ax[1].step(k, X.cdf(k), where="post", color="green")
ax[1].set_title("CDF Poisson(λ=4)")
ax[1].set_xlabel("k")
ax[1].set_ylabel("P(X <= k)")

plt.tight_layout()
plt.savefig("tahap3_poisson.png", dpi=150)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

p = 0.25
X = stats.geom(p)      # scipy: nilai mulai dari k = 1, 2, 3, ...

print("P(X=3)  :", round(X.pmf(3), 6))
print("P(X<=3) :", round(X.cdf(3), 6))
print("Mean    :", X.mean())
print("Varians :", X.var())

sim = X.rvs(size=10000, random_state=42)
print("Mean simulasi    :", round(sim.mean(), 3))
print("Varians simulasi :", round(sim.var(), 3))

k = np.arange(1, 21)
frek_sim = np.array([(sim == i).mean() for i in k])

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
lebar = 0.4
ax[0].bar(k - lebar/2, X.pmf(k), width=lebar, label="Teori (pmf)", color="steelblue")
ax[0].bar(k + lebar/2, frek_sim, width=lebar, label="Simulasi", color="orange")
ax[0].set_title("Geometrik(p=0.25): pmf vs simulasi")
ax[0].set_xlabel("k (percobaan pertama yang berhasil)")
ax[0].set_ylabel("Peluang")
ax[0].legend()

ax[1].step(k, X.cdf(k), where="post", color="green")
ax[1].set_title("CDF Geometrik(p=0.25)")
ax[1].set_xlabel("k")
ax[1].set_ylabel("P(X <= k)")

plt.tight_layout()
plt.savefig("tahap3_geometrik.png", dpi=150)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

a, b = 100, 300
X = stats.uniform(loc=a, scale=b - a)    # scipy: loc = batas bawah, scale = lebar (b-a)

print("f(150)          :", round(X.pdf(150), 4))
print("P(150<=X<=250)  :", round(X.cdf(250) - X.cdf(150), 4))
print("P(X>280)        :", round(X.sf(280), 4))
print("Mean            :", round(X.mean(), 4))
print("Varians         :", round(X.var(), 4))

sim = X.rvs(size=10000, random_state=42)
print("Mean simulasi    :", round(sim.mean(), 3))
print("Varians simulasi :", round(sim.var(), 3))

x = np.linspace(80, 320, 500)
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(sim, bins=30, stat="density", color="lightgray", ax=ax[0], label="Simulasi")
ax[0].plot(x, X.pdf(x), color="red", label="pdf teori")
ax[0].set_title("Uniform(100, 300): pdf vs simulasi")
ax[0].set_xlabel("Waktu respons (ms)")
ax[0].set_ylabel("Kepadatan")
ax[0].legend()

ax[1].plot(x, X.cdf(x), color="green")
ax[1].set_title("CDF Uniform(100, 300)")
ax[1].set_xlabel("Waktu respons (ms)")
ax[1].set_ylabel("P(X <= x)")

plt.tight_layout()
plt.savefig("tahap3_uniform.png", dpi=150)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

mu, sigma = 120, 15
X = stats.norm(loc=mu, scale=sigma)

print("f(120)          :", round(X.pdf(120), 4))
print("P(X<=100)       :", round(X.cdf(100), 4))
print("P(105<=X<=135)  :", round(X.cdf(135) - X.cdf(105), 4))
print("Persentil ke-95 :", round(X.ppf(0.95), 2))
print("Mean            :", X.mean())
print("Varians         :", X.var())

sim = X.rvs(size=10000, random_state=42)
print("Mean simulasi    :", round(sim.mean(), 3))
print("Varians simulasi :", round(sim.var(), 3))
print("Proporsi 105-135 :", round(((sim >= 105) & (sim <= 135)).mean(), 4))

x = np.linspace(60, 180, 500)
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(sim, bins=40, stat="density", color="lightgray", ax=ax[0], label="Simulasi")
ax[0].plot(x, X.pdf(x), color="red", label="pdf teori")
x_area = np.linspace(105, 135, 200)
ax[0].fill_between(x_area, X.pdf(x_area), color="skyblue", alpha=0.5, label="P(105<=X<=135)")
ax[0].set_title("Normal(120, 15): pdf vs simulasi")
ax[0].set_xlabel("Latensi (ms)")
ax[0].set_ylabel("Kepadatan")
ax[0].legend()

ax[1].plot(x, X.cdf(x), color="green")
ax[1].set_title("CDF Normal(120, 15)")
ax[1].set_xlabel("Latensi (ms)")
ax[1].set_ylabel("P(X <= x)")

plt.tight_layout()
plt.savefig("tahap3_normal.png", dpi=150)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

lam = 2
X = stats.expon(scale=1/lam)     # scipy: scale = 1/λ

print("f(0.5)    :", round(X.pdf(0.5), 4))
print("P(X<=0.5) :", round(X.cdf(0.5), 4))
print("P(X>1)    :", round(X.sf(1), 4))
print("Mean      :", X.mean())
print("Varians   :", X.var())

sim = X.rvs(size=10000, random_state=42)
print("Mean simulasi    :", round(sim.mean(), 4))
print("Varians simulasi :", round(sim.var(), 4))
print("Proporsi X>1 simulasi:", round((sim > 1).mean(), 4))

x = np.linspace(0, 3, 500)
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(sim, bins=40, stat="density", color="lightgray", ax=ax[0], label="Simulasi")
ax[0].plot(x, X.pdf(x), color="red", label="pdf teori")
ax[0].set_xlim(0, 3)
ax[0].set_title("Exponential(λ=2): pdf vs simulasi")
ax[0].set_xlabel("Waktu antar paket (detik)")
ax[0].set_ylabel("Kepadatan")
ax[0].legend()

ax[1].plot(x, X.cdf(x), color="green")
ax[1].set_title("CDF Exponential(λ=2)")
ax[1].set_xlabel("Waktu antar paket (detik)")
ax[1].set_ylabel("P(X <= x)")

plt.tight_layout()
plt.savefig("tahap3_exponential.png", dpi=150)
plt.show()
```
